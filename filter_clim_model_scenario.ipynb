```python
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', 60)
pd.set_option('display.max_rows', 100)
```

```python
path_list_scenario = "data/pangeo-cmip6.csv"
```

```python
scenario = pd.read_csv(path_list_scenario)
```

```python
activities = ["CMIP", "ScenarioMIP"]
experiments = ["historical", "ssp370", "ssp585"]
tables = ["Amon", "Omon"]
variables = ["hurs", "psl", "ta", "tos"]
```

```python
scenario_filter = scenario[
    scenario["activity_id"].isin(activities) &
    scenario["experiment_id"].isin(experiments) &
    scenario["table_id"].isin(tables) &
    scenario["variable_id"].isin(variables) &
    (scenario["grid_label"] == "gn")
]
```

```python
scenario_filter
```

```python
scenario_filter["Nb_source_member_diff"] = scenario_filter.groupby(["source_id", "member_id"])["activity_id"].transform(lambda x: x.size)
```

```python
scenario_filter = scenario_filter[scenario_filter["Nb_source_member_diff"] == 12].reset_index(drop=True).drop(columns=["Nb_source_member_diff"])
```

```python
scenario_filter.to_excel("data/pangeo-cmip6-filtered.xlsx", index=False)
```

```python
scenario_filter.member_id.value_counts()
```

```python
scenario["Scenario_exist"] = (
    scenario
    .groupby("source_id")["experiment_id"]
    .transform(lambda x: experiments.issubset(set(x)))
)

scenario["Table_exist"] = (
    scenario
    .groupby("source_id")["table_id"]
    .transform(lambda x: tables.issubset(set(x)))
)

scenario["Variable_exist"] = (
    scenario
    .groupby("source_id")["variable_id"]
    .transform(lambda x: variables.issubset(set(x)))
)
```

```python
scenario.head()
```

```python
scenario = scenario[scenario["activity_id"].isin(activities) & scenario["Scenario_exist"] & scenario["Table_exist"] & scenario["Variable_exist"] & (scenario["grid_label"]=="gn")]
```
